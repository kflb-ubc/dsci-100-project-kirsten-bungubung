#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
### EDIT THIS CODE CELL ###
mean_min_max <- summarize(my_data, 
                          mean_conf_wins_pct = mean(conf_wins_pct, na.rm = TRUE), #mean, min and max of both numeric variables calculated independently
                          min_conf_wins_pct = min(conf_wins_pct, na.rm = TRUE), #accounting for NAs
                          max_conf_wins_pct = max(conf_wins_pct, na.rm = TRUE),
                          mean_reg_wins_pct = mean(reg_wins_pct), #no NAs in the reg_wins_pct column, so no need to add "na.rm = TRUE"
                          min_reg_wins_pct = min(reg_wins_pct), 
                          max_reg_wins_pct = max(reg_wins_pct)) 
mean_min_max
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
### EDIT THIS CODE CELL ###
na_conf <- my_data |>
                filter(is.na(conf_wins_pct)) |> #only the rows with NA for one variable
                nrow() |> #get the number of rows with NA
                as_tibble() |> #making a 1x1 tibble with the number of NAs for this variable
                rename(conf_wins_pct = value) #renaming the column to the variable name
                
na_reg <- my_data |>
                filter(is.na(reg_wins_pct)) |>
                nrow() |>
                as_tibble() |>
                rename(reg_wins_pct = value)

na_finish <- my_data |>
                filter(is.na(tourney_finish)) |>
                nrow() |>
                as_tibble() |>
                rename(tourney_finish = value)

missingness <- bind_cols(na_conf, na_reg, na_finish) #combining my 1x1 tibbles with the pre-named, pre-calculated values for NA count per variable
missingness
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.3: Summarize over levels of a categorical variable (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates mean of one numeric variable for each of the levels of another categorical variable. 
- Show the outputted table. 
</div>

```r
### EDIT THIS CODE CELL ###
mean <- my_data |>
            group_by(tourney_finish) |> #specifying that the means of my numeric variable should be grouped by the values of my categorical variable
            summarize(mean_reg_wins_pct = mean(reg_wins_pct)) |> #taking the means of my numeric variable
            select(mean_reg_wins_pct, tourney_finish) #selecting for only the categorical and means of the numeric variables
mean
```
